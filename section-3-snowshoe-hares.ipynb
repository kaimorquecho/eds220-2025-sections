# Snowshoe hares at Bonanza Creek Experimental Forest

### Data Description: 
The dataset is the result of a study examining snowshoe hare (Lepus americanus) populations (a keystone species in northern boreal forests) across five sites in Alaska’s Tanana Valley from 1999 to 2002. Using capture-recapture methods, researchers found that hare densities peaked in 1999 and declined in subsequent years. Despite the overall population decline, apparent survival rates did not decrease, and movement patterns remained consistent over time. Persistence of individuals during population declines appeared linked to better body condition at peak density. Additionally, researchers found only a weak relationship between hare pellet counts and actual population density, suggesting that pellet-based estimates may be unreliable for assessing hare densities in Interior Alaska.
